# Demystify the API: making requests in code

The code from the *Making Requests in Code* lesson.

**Before you run anything**
1. Create a free [Resend](https://resend.com/signup) account and an API key.
2. Click the **key icon** in the left sidebar (Secrets), add a secret named `resend_api_key`, paste your key as the value and switch on **Notebook access**.
3. Replace `you@example.com` with the email you signed up to Resend with. On the free plan, the test sender only delivers to that address.

Never paste your API key into the code itself.

## 1. Send one email
The code Postman exported, with the key loaded from Colab Secrets.

```python
import requests
import json
from google.colab import userdata

my_api_key = userdata.get('resend_api_key')

url = "https://api.resend.com/emails"

payload = json.dumps({
  "from": "onboarding@resend.dev",
  "to": "you@example.com",
  "subject": "My first Python API email",
  "html": "<p>I just sent this email with a Python API!</p>"
})
headers = {
  'Content-Type': 'application/json',
  'Authorization': f'Bearer {my_api_key}'
}

response = requests.request("POST", url, headers=headers, data=payload)

print(response.status_code)
print(response.text)
```

## 2. Send a personalized email to each name
Only edit `your_email` and `names`. `time.sleep(1)` waits a second between emails because Resend's free plan allows 2 requests per second.

```python
import requests, time
from google.colab import userdata

api_key = userdata.get("resend_api_key")   # from Colab Secrets
your_email = "you@example.com"             # the email you signed up to Resend with
names = ["Anna", "Ben", "Chloe"]

url = "https://api.resend.com/emails"
headers = {"Authorization": f"Bearer {api_key}", "Content-Type": "application/json"}

for name in names:
    payload = {
        "from": "onboarding@resend.dev",
        "to": your_email,
        "subject": f"Hi {name}, this email came from Python",
        "html": f"<p>Hey {name}, I just sent this with the Resend API!</p>",
    }
    response = requests.post(url, json=payload, headers=headers)
    print(name, response.status_code, response.text)
    time.sleep(1)  # free plan: max 2 requests per second
```
